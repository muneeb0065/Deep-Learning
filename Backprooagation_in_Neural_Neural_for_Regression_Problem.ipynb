```python
import numpy as np
import pandas as pd
```

```python
data = pd.DataFrame([[8,8,4], [7,9,5], [6,10,6], [5,12,7]], columns= ['cgpa', 'profile_score', 'lpa'])
```

```python
def initialize_parameters(layer_dims):

  # np.random.seed(3)  #  this is used when we are actually generating the random weights for our neural networks. It mean every time the block run it generates the random numbers from the same page which contains a lot of random numbers. and here is the page number is 3.
  parameters = {}
  for l in range(1, len(layer_dims)): # As the 1st layer is input layer thats why we start from 1 instead of writing only len(layers_dims)

    parameters['W' + str(l)] = np.ones((layer_dims[l-1], layer_dims[l]))*0.1  # This piece is written in a way that it generates the weights and bias when we provide the no of nodes is each layer in the form of a list.
    parameters['b' + str(l)] = np.zeros((layer_dims[l], 1))


  return parameters
```

```python
initialize_parameters([2,2,1])
# print(parameters['W1'].shape)
# print(parameters['W1'])
```

```python
# Now this is the function that calculates the output z of any neuron

def linear_forward(A_prev, W, b):
  z = np.dot(W.T,A_prev) +b

  return z
```

```python
# Now this is the code for making predictions (Forward Propagation)

def forward_prop(X, parameters):
  A = X
  L = len(parameters)//2

  for l in range(1,L+1): # As we have 2 layers so this loop runs 2 times
    A_prev = A
    Wl = parameters['W'+str(l)]
    bl = parameters['b'+str(l)]
    # print("A"+str(l-1)+": ", A_prev)
    # print(A_prev.shape)
    # print("W"+str(l)+": ", Wl)
    # print("b"+str(l)+": ", bl)
    # print("--"*20)

    A = linear_forward(A_prev, Wl, bl)
    # print("A"+str(l)+": ", A)
    # print("**"*20)

  return A,A_prev # I am also returning A_prev (outputs of previous layer from output layer) because when applying the backpropagation i need them while calculataing the derivative.

# If face any issue just take copy pen and understand it, it's quite simple
```

```python
X = data[['cgpa', 'profile_score']].values[0].reshape(2,1) # Shape(no of features, no. of training example)
y = data[['lpa']].values[0][0] # here [0][0] means to get the value not in the form of array

# Parameter initialization
parameters = initialize_parameters([2,2,1])

y_hat,A1 = forward_prop(X, parameters)

print(y_hat[0][0],A1)
```

```python
# This is the code for updating all the parameteres

# Wnew = Wold - lr(gradient of Loss w.r.t to all the weights and bias)

def update_parameters(parameters,y,y_hat,A1,X):
  parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * 2 * (y - y_hat)*A1[0][0])
  parameters['W2'][1][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat)*A1[1][0])
  parameters['b2'][0][0] = parameters['W2'][1][0] + (0.001 * 2 * (y - y_hat))

  parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][0][0])

  parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * (y - y_hat)*parameters['W2'][1][0])
```

```python
# These are the updated parameters

parameters
```

```python
# Now i will make predictions for 2nd row

X = data[['cgpa', 'profile_score']].values[1].reshape(2,1)
y = data[['lpa']].values[1][0]

y_hat,A1 = forward_prop(X,parameters)
y_hat = y_hat[0][0]


# After calculating the prediction i will calculate the loss and again update the weights


update_parameters(parameters,y,y_hat,A1,X)

parameters
```

```python
# Now instead of doing this manually we will create a loop for the number of epochs we want and do the same process for given number of epochs

# Initialize the params
parameters = initialize_parameters([2,2,1])
epochs = 10

for i in range(epochs):

  Loss = []

  for j in range(data.shape[0]):

    X = data[['cgpa', 'profile_score']].values[j].reshape(2,1) # Shape(no of features, no. of training example)
    y = data[['lpa']].values[j][0]

    # Parameter initialization


    y_hat,A1 = forward_prop(X,parameters)
    y_hat = y_hat[0][0]

    update_parameters(parameters,y,y_hat,A1,X)

    Loss.append((y-y_hat)**2)

  print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())

parameters
```
