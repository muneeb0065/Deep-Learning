#**LeNet Architecture**
**This is the first cnn model developed by Yan Lecun in 1998 and its realworld use is for automated reading of US Postal Service ZIP codes and bank checks/ATMs.**

In this architecture Yan used two conv layers:

1. In 1st layer there are 6 filters of size 5x5 and used the average pooling with size 2,2 and the stride is also 2.

2. In 2nd layer there are 16 filters of size 5x5 and same pooling as in above layer

Then the rest of 3 layers are the Fully Connected Layers with details:

3. The 3rd layer has 120 neurons that recieves the input after the flatenning the tensor.

4. The 4th layer has 84 neurons. with actiation of tanh.

5. The 5th layer has 10 neurons and activation is softmax because there are total 10 digits.

```python
model = Sequential()

model.add(Conv2D(6,kernel_size=(5,5), padding='valid', activation='tanh', input_shape=(32,32,1))) # look at that time ReLU was not discovered so Dr. Yan used the tanh because this was the best activation at that time.
model.add(AveragePooling2D(pool_size=(2,2), strides = 2, padding = 'valid'))

model.add(Conv2D(16,kernel_size=(5,5), padding='valid', activation='tanh'))
model.add(AveragePooling2D(pool_size=(2,2), strides = 2, padding = 'valid'))

model.add(Flatten())

model.add(Dense(120, activation='tanh'))
model.add(Dense(84, activation='tanh'))
model.add(Dense(10, activation='softmax'))
```
