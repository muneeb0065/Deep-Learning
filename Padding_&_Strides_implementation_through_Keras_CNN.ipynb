```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

#**Padding**

```python
model = Sequential()

# Just like dense layers in ANN, we have Conv2D for convolution operations.

# Look we have added the pading as same means we dont want to perform padding and perform normal convolutional operation in which the feature map size will be reduced.
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))

# We are converting this 2d array of pixels into 1D so that we can easily feed as input.
model.add(Flatten())

# After performing the convolutional operation we are making a dense layers network.
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()

# So looking at the summary you can clearly see the feature map size is reducing as normal.
```

###**Now we will apply padding.**

```python
model = Sequential()


# We just use the padding value as same and keras will automatically apply best padding
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same', activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
# Now if you see the summary you can see that the size of our resultating image is not reducing and we are not losing any information.
model.summary()
```

#**Strides**

```python
model = Sequential()


# Here you can see that we are applying the stride of size 2,2
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu', input_shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',strides=(2,2), activation='relu'))

model.add(Flatten())

model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()

# Now you can see that when we add the the feature map size will be redued significantly.
```
